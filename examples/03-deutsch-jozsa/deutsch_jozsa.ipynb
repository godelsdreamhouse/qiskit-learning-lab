```python
#micah gentry
#Deutsch-Jozsa algorithm: is f balanced or constant?
# balanced means f is 0 exactly as often as it is 1
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
import pandas as pd

print("is f balanced or constant?")
print("if f(0)=f(1), its constant and if f(0)!=f(1), its balanced")
#initialize a quantum circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)
#flip q1 from |0> to |1>
qc.x(1) #now our starting state is |01>
#apply hadamard gates
qc.h(0) #|+>
qc.h(1) #|->

#the oracle
#four possible boolean functions f:{0,1} -> {0,1}
print("you must choose an oracle to continue")
print("1. f(x)=0")
print("2. f(x)=1")
print("3. f(x)=x")
print("4. f(x)=!x")
n = int(input("enter corresponding int: "))

while n < 1 or n > 4:
  n = int(input("invalid input. try again: "))

if n == 2:
  qc.x(1)

if n == 3:
  qc.cx(0,1)

if n == 4:
  qc.x(0)
  qc.cx(0,1)
  qc.x(0)

qc.h(0)

qc.measure(0, 0)

#make a simulator
sim = AerSimulator()

#store the results of the simulation
results = sim.run(qc, shots=1024).result()
counts = results.get_counts()

outcome = max(counts, key=counts.get)

q0_result = outcome[-1]

if q0_result == "0":
  print("the function is constant")
else:
  print("the function is balanced")

print(qc.draw())
```
