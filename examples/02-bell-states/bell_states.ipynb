```python
#micah gentry
#bell states

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

circuits = [QuantumCircuit(2, 2) for _ in range(4)]
qc1, qc2, qc3, qc4 = circuits

#B_00 = (|00> + |11>)/ sqr(2)

qc1.h(0) #takes q0 from |0> to a superposition of 0 and 1
qc1.cx(0, 1) # apply cnot with q0 as control and q1 as target
qc1.measure([0, 1], [0, 1]) #measure q0, q1 in terms of 0 and 1
print(qc1.draw()) #draw our circuit

#B_01 = (|00> - |11>)/ sqr(2)

qc2.h(0)
qc2.cx(0, 1)
qc2.z(0)
qc2.measure([0,1], [0,1])
print(qc2.draw())

#B_10 = (|01> + |10>)/ sqr(2)

qc3.h(0)
qc3.cx(0,1)
qc3.x(1)
qc3.measure([0,1], [0,1])
print(qc3.draw())

#B_11 = (|01> - |10>)/ sqrt(2)

qc4.h(0)
qc4.cx(0,1)
qc4.x(1)
qc4.z(0)
qc4.measure([0,1],[0,1])
print(qc4.draw())


```
